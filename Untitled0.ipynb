```python

# =====================================
# SIMPLE AI CHATBOT USING PYTHON
# =====================================

!pip -q install gradio

import gradio as gr
import random

# Chatbot response function
def chatbot(message, history):

    msg = message.lower().strip()

    if any(word in msg for word in ["hello", "hi", "hey"]):
        return "Hello! Welcome. How can I help you today?"

    elif "your name" in msg:
        return "I am SmartBot, your AI chatbot assistant."

    elif "how are you" in msg:
        return "I'm doing great! What would you like to know?"

    elif "rag" in msg:
        return "RAG stands for Retrieval-Augmented Generation. It retrieves relevant information from documents before generating an answer."

    elif "ollama" in msg:
        return "Ollama is a tool used to run large language models locally."

    elif "python" in msg:
        return "Python is a popular programming language used in AI, machine learning and web development."

    elif "ai" in msg:
        return "Artificial Intelligence enables machines to perform tasks that normally require human intelligence."

    elif "bye" in msg:
        return "Goodbye! Have a wonderful day!"

    elif "thank" in msg:
        return "You're welcome! Happy to help."

    else:
        responses = [
            "That's an interesting question! Can you explain a little more?",
            "I am SmartBot. I am here to help you.",
            "Could you try asking that in a different way?",
            "Let me know what topic you want to explore."
        ]
        return random.choice(responses)


# Create chatbot interface
demo = gr.ChatInterface(
    fn=chatbot,
    title="🤖 SmartBot - AI Chatbot",
    description="Welcome to SmartBot! Ask me anything.",
    examples=[
        "Hello",
        "What is RAG?",
        "What is Python?",
        "What is Artificial Intelligence?",
        "What is Ollama?"
    ]
)

# Launch chatbot
demo.launch(share=True)
```
